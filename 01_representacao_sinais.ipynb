SINAIS DISCRETOS

```python
x=[1,0,3,2]
h=[1,1,1]
def convolve_direct(x, h):
    N = len(x)
    M = len(h)
    L = N+M-1                      # comprimento da saida (formula que acabamos de ver)
    y = np.zeros(L)                # vetor de saida, comeca zerado

    for n in range(L):
        soma = 0
        for k in range(N):
            idx_h = n-k           # qual indice de h corresponde a esse (n, k)?
            if 0 <= idx_h < M:      # só soma se for um índice válido de h
                soma = soma + x[k] * h[idx_h]
        y[n] = soma

    return y

y = convolve_direct(x, h)
print(y)
```
